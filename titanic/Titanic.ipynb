## O Danych

Dane o pasażerach Titanica

Zbiór danych zawiera informacje o pasażerach RMS Titanic, który zatonął 15 kwietnia 1912 roku po zderzeniu z górą
lodową. Dane obejmują takie atrybuty jak klasa podróży, wiek, płeć, liczba rodzeństwa/małżonków na pokładzie,
liczba rodziców/dzieci na pokładzie, cena biletu oraz miejsce zaokrętowania.

Zbiór zawiera także informację o tym, czy pasażer przeżył katastrofę.

Titanic przewoził ponad 2,200 osób, z czego ponad 1,500 zginęło, co czyni tę katastrofę jedną z najbardziej
tragicznych w historii morskiej.

Kolumny:

* **pclass** - Klasa biletu
* **survived** - Czy pasażer przeżył katastrofę
* **name** - Imię i nazwisko pasażera
* **sex** - Płeć pasażera
* **age** - Wiek pasażera
* **sibsp** - Liczba rodzeństwa/małżonków na pokładzie
* **parch** - Liczba rodziców/dzieci na pokładzie
* **ticket** - Numer biletu
* **fare** - Cena biletu
* **cabin** - Numer kabiny
* **embarked** - Port, w którym pasażer wszedł na pokład (C = Cherbourg, Q = Queenstown, S = Southampton)
* **boat** - Numer łodzi ratunkowej
* **body** - Numer ciała (jeśli pasażer nie przeżył i ciało zostało odnalezione)
* **home.dest** - Miejsce docelowe

## Oblicz jaki procent spośród wszystkich analizowanych rekordów stanowią pasażerowie, którzy przeżyli katastrofę

## Procentowy udział pasażerów pod względem płci oraz grupy wiekowej w analizownym zbiorze danych
### A) Wypełniając brakujące informacje na temat wieku pasażera przy użyciu mediany dla tej wartości (pośród analizowanego zbioru danych)

```python
import matplotlib.pyplot as plt
df_filled = df.copy()

# Fill missing values in 'age' with the median age
df_filled['age'].fillna(df['age'].median(), inplace=True)

# Create age groups
df_filled['age_group'] = df_filled['age'].apply(lambda x: '< 18' if x < 18 else '>= 18')

# Group by sex and age group
grouped = df_filled.groupby(['sex', 'age_group']).size().unstack(fill_value=0)

# Calculate percentages
total_passengers = df_filled['name'].notna().sum()
percentages = (grouped / total_passengers) * 100

# Plotting
ax = percentages.plot(kind='bar', stacked=True)
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%')

plt.title('Percentage of Passengers by Gender and Age Group')
plt.xlabel('Gender')
plt.ylabel('Percentage')
plt.xticks(rotation=0)
plt.legend(title='Age Group')
plt.tight_layout()

# Save the plot
plot_filename = "passenger_distribution.png"
plt.savefig(plot_filename)

# Declare result var
result = { "type": "plot", "value": plot_filename }
```

### B) Pomijając rekordy z brakującymi informacjami na temat wieku pasażera

```python
# Group by 'pclass' and count the number of passengers
passenger_counts = df['pclass'].value_counts().sort_index()

# Create a bar plot
fig, ax = plt.subplots()
bars = ax.bar(passenger_counts.index, passenger_counts.values)

# Add labels above each bar
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height}', xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),  # 3 points vertical offset
                textcoords="offset points",
                ha='center', va='bottom')

# Set title and labels
ax.set_title('Number of Passengers by Ticket Class')
ax.set_xlabel('Ticket Class')
ax.set_ylabel('Number of Passengers')

# Increase the top margin to ensure the labels are not cut off
plt.ylim(0, max(passenger_counts.values) * 1.1)

# Save the plot as a PNG file
plot_filename = "passenger_counts.png"
```

## Liczba pasażerów z uwzględnieniem podziału na płeć i klasę biletu

```python
# Drop rows with missing values in 'pclass' or 'sex'
df = df.dropna(subset=['pclass', 'sex'])

# Group by 'pclass' and 'sex' and count the number of passengers
passenger_counts = df.groupby(['pclass', 'sex']).size().unstack()

# Plotting the data
ax = passenger_counts.plot(kind='bar', stacked=False, figsize=(10, 6))

# Adding labels and title
ax.set_xlabel('Class')
ax.set_ylabel('Number of Passengers')
ax.set_title('Number of Passengers by Class and Gender')

# Adding exact values on top of the bars
for container in ax.containers:
    ax.bar_label(container)

# Adjust layout to make room for labels
plt.tight_layout()

# Save the plot as a PNG file
plot_filename = "passenger_counts_by_class_and_gender.png"
plt.savefig(plot_filename)

# Declare result variable
result = { "type": "plot", "value": plot_filename }
```

```python

# Drop rows with missing values in 'pclass' or 'sex'
df_cleaned = df.dropna(subset=['pclass', 'sex'])

# Calculate the total number of passengers
total_passengers = len(df_cleaned)

# Group by 'pclass' and 'sex' and count the number of passengers in each group
grouped = df_cleaned.groupby(['pclass', 'sex']).size().reset_index(name='count')

# Calculate the percentage share
grouped['percentage'] = (grouped['count'] / total_passengers) * 100

# Plotting the results
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(
    grouped.apply(lambda row: f"{int(row['pclass'])} - {row['sex']}", axis=1),
    grouped['percentage']
)

# Adding exact numbers on top of the bars
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height:.2f}%', xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3), textcoords="offset points", ha='center', va='bottom')

ax.set_xlabel('Class - Gender')
ax.set_ylabel('Percentage of Total Passengers')
ax.set_title('Percentage Share of Total Passengers by Class and Gender')

# Save the plot as a PNG file
plot_filename = "percentage_share_by_class_and_gender.png"
plt.savefig(plot_filename)

# Declare result variable
result = { "type": "plot", "value": plot_filename }
```

## Odsetek osób, które przeżyły posiadając bilet danej klasy

```python
# Drop rows with missing values in 'pclass' or 'survived' columns
df = df.dropna(subset=['pclass', 'survived'])

# Calculate the percentage of survivors for each class
survival_rate = df.groupby('pclass')['survived'].mean() * 100

# Plotting the survival rate by class
fig, ax = plt.subplots()
bars = ax.bar(survival_rate.index, survival_rate)

# Adding exact values above the bars
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height:.2f}%', 
                xy=(bar.get_x() + bar.get_width() / 2, height), 
                xytext=(0, 3),  # 3 points vertical offset
                textcoords="offset points", 
                ha='center', va='bottom')

# Adding title and labels
ax.set_title('Percentage of Survivors by Ticket Class')
ax.set_xlabel('Ticket Class')
ax.set_ylabel('Survival Rate (%)')

# Adjusting the top margin to ensure no text is cut off
plt.ylim(0, max(survival_rate) + 10)

# Save the plot as a PNG file
plot_filename = "survival_rate_by_class.png"
plt.savefig(plot_filename)

# Declare result variable
result = { "type": "plot", "value": plot_filename }
```

## Odsetek osób, które przeyżyły posiadając bilet danej klasy z uwzględnieniem podziału ze względu na płeć pasażera

## Liczba osób, które zginęły z uwzględnieniem podziału ze względu na klasę biletu oraz płeć

```python
# Filter out rows where 'survived' is NaN
df = df.dropna(subset=['survived'])

# Calculate the number of people who died for each class and gender
death_counts = df[df['survived'] == 0].groupby(['pclass', 'sex']).size().unstack()

# Plotting the bar chart
ax = death_counts.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.title('Number of People Who Died by Class and Gender')
plt.xlabel('Ticket Class')
plt.ylabel('Number of Deaths')
plt.xticks(rotation=0)

# Adding exact values on top of the bars
for p in ax.patches:
    ax.annotate(str(int(p.get_height())), (p.get_x() * 1.005, p.get_height() * 1.005))

# Save the plot as a PNG file
plot_filename = "death_counts_by_class_and_gender.png"
plt.savefig(plot_filename)

# Declare result variable
result = { "type": "plot", "value": plot_filename }
```

## Konklzuja
Każda z tych wartości powinna odpowiadc procentowemu udziałowi w liczbie osób które zginęły, który zostanie przedtawiony poniżej

```python
# Filter out rows where 'survived' or 'sex' or 'pclass' is NaN
df = df.dropna(subset=['survived', 'sex', 'pclass'])

# Calculate the percentage of people who died for each class and gender
death_counts = df[df['survived'] == 0].groupby(['pclass', 'sex']).size()
total_counts = df.groupby(['pclass', 'sex']).size()
death_percentage = (death_counts / total_counts * 100).unstack()

# Plotting the results
ax = death_percentage.plot(kind='bar', stacked=False, figsize=(10, 6))
plt.title('Percentage of People Who Died by Ticket Class and Gender')
plt.xlabel('Ticket Class')
plt.ylabel('Percentage (%)')
plt.ylim(0, 100)

# Adding exact values on top of the bars
for p in ax.patches:
    ax.annotate(f'{p.get_height():.1f}%', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                xytext=(0, 9), 
                textcoords='offset points')

# Adjusting layout to ensure everything fits
plt.tight_layout()

# Save the plot as a PNG file
plot_filename = "death_percentage_by_class_and_gender.png"
plt.savefig(plot_filename)

# Declare result variable
result = { "type": "plot", "value": plot_filename }
```

## Konkluzje
1. Dane te wydają się spójne z danymi procentowymi dla osób, które przeżyły katastrofę.
2. Każdej wartości procentowej można przyporządkować konkretną wartość liczbową. Przykładowo 3,5% kobiet z pierwszej klasy, które zginęły odpowiada 5 zmarłym kobietom. Analogicznie działa to dla każdej z kolumn.
3. Wartości te można porównać z całkowitą liczbą pasażerów dla każdej z klas.

## Liczba osób poniżej 18 roku życia, które przeżyły ktastrofę

```python
# Filter for people below 18 years of age
children_df = df[df['age'] < 18]

# Calculate total number of children
total_children = len(children_df)

# Calculate the number of children who survived
survived_children = children_df[children_df['survived'] == 1]

# Calculate the percentage of children who survived
percentage_survived = (len(survived_children) / total_children) * 100

# Group by gender and ticket class
grouped = children_df.groupby(['sex', 'pclass']).agg({'survived': 'sum', 'name': 'count'}).reset_index()
grouped.columns = ['sex', 'pclass', 'survived', 'total']

# Plotting the data
fig, ax = plt.subplots(figsize=(10, 6))

# Create a bar plot
grouped.pivot(index='pclass', columns='sex', values='total').plot(kind='bar', ax=ax)

# Add exact numbers above the bars
for p in ax.patches:
    ax.annotate(f'{p.get_height()}', (p.get_x() * 1.005, p.get_height() * 1.005))

plt.title('Number of Children Below 18 Years Who Survived by Gender and Ticket Class')
plt.xlabel('Ticket Class')
plt.ylabel('Number of Children')
plt.legend(title='Gender')
plt.tight_layout()

# Save the plot
plot_filename = 'children_survival_by_gender_class.png'
plt.savefig(plot_filename)

# Prepare the result dictionary
result = {
    "type": "plot",
    "value": plot_filename
}
```

```python
# Filter data for passengers below 18 years of age who survived
df_under_18_survived = df[(df['age'] < 18) & (df['survived'] == 1)]

# Group by class and sex, then calculate the percentage
grouped = df_under_18_survived.groupby(['pclass', 'sex']).size().reset_index(name='count')
total_by_class_sex = df[df['age'] < 18].groupby(['pclass', 'sex']).size().reset_index(name='total')
merged = pd.merge(grouped, total_by_class_sex, on=['pclass', 'sex'])
merged['percentage'] = (merged['count'] / merged['total']) * 100

# Plotting
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(merged.index, merged['percentage'], tick_label=merged.apply(lambda x: f"{int(x['pclass'])} {x['sex']}", axis=1))

# Adding percentage values on top of the bars
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height:.2f}%', xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3), textcoords="offset points", ha='center', va='bottom')

# Adding title and labels
ax.set_title('Percentage of Survivors Below 18 Years by Class and Gender')
ax.set_xlabel('Class and Gender')
ax.set_ylabel('Percentage')

# Save the plot as a PNG file
plt.tight_layout()
plt.savefig("survivors_below_18_by_class_gender.png")

# Declare result var
result = { "type": "plot", "value": "survivors_below_18_by_class_gender.png" }
```
